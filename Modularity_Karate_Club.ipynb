### Abstract
This notebook implements **Newman's spectral modularity method** for community detection from first principles on Zachary's classic Karate Club network. We:

1. Construct the modularity matrix $B = A - \frac{\mathbf{k}\mathbf{k}^\top}{2m}$
2. Perform recursive spectral bipartition guided by the leading eigenvector
3. Stop splitting a community when it can no longer be divided (leading eigenvalue $\lambda_1^{(C)} \le 0$, or the leading eigenvector cannot separate the nodes — see Section 4)
4. Visualize the evolving partition after every split
5. Track degree, betweenness, closeness centrality and clustering coefficient of every node after every split
6. Discuss which nodes remain structurally central and how community structure shapes the metrics

All mathematics follows Newman (2006) *"Modularity and community structure in networks"*, PNAS 103(23):8577–8582.

## 3. Mathematics – The Modularity Matrix

For an undirected graph $G=(V,E)$ with $n=|V|$ vertices and $m=|E|$ edges, the adjacency matrix $A$ and degree vector $\mathbf{k}$ yield the **modularity matrix**

$$
B = A - \frac{\mathbf{k}\mathbf{k}^\top}{2m}.
$$

Entry $B_{ij}$ measures how much the observed edge $A_{ij}$ exceeds the expected number of edges $k_ik_j/2m$ under a configuration-model null that preserves degrees.

The modularity of a bipartition encoded by $s\in\{-1,+1\}^n$ is

$$
Q = \frac{1}{4m}\, s^\top B s.
$$

By the Rayleigh–Ritz theorem the continuous relaxation $\max_{\|s\|_2=1}s^\top Bs$ is solved by the leading eigenvector $u_1$ of $B$. Rounding the signs of $u_1$ produces a discrete bipartition.

When a community $C\subseteq V$ is already isolated we form the *restricted* matrix $B^{(C)}$ by extracting the corresponding rows/columns of the *global* $B$ (the null model stays global), exactly as in the assignment handout.

**Helper functions** (the sign of an eigenvector is arbitrary, so we fix it by making the largest-magnitude entry positive – this makes the notebook fully reproducible across machines).

```python
def modularity_matrix(G, nodelist=None):
    '''Construct the modularity matrix B = A - kk^T/(2m) for an undirected graph.'''
    if nodelist is None:
        nodelist = sorted(G.nodes())
    A = nx.to_numpy_array(G, nodelist=nodelist, weight=None)
    k = A.sum(axis=1)
    m = A.sum() / 2.0
    if m == 0:
        return np.zeros_like(A), nodelist, 0.0
    B = A - np.outer(k, k) / (2.0 * m)
    B = (B + B.T) / 2.0
    return B, nodelist, m


def leading_eigenpair(M):
    '''Largest eigenvalue of a real symmetric matrix and its unit eigenvector (sign-fixed).'''
    evals, evecs = np.linalg.eigh(M)          # eigenvalues in ascending order
    lam, u = evals[-1], evecs[:, -1].copy()
    u *= np.sign(u[np.argmax(np.abs(u))])     # eigenvectors are only defined up to +-1
    return lam, u


def modularity_of_partition(B, m, nodes, partition):
    '''Q = (1/2m) * sum_ij B_ij * delta(c_i, c_j)   (multi-community version of Q = s^T B s / 4m).'''
    idx = {v: i for i, v in enumerate(nodes)}
    label = np.empty(len(nodes), dtype=int)
    for c, comm in enumerate(partition):
        for v in comm:
            label[idx[v]] = c
    same = label[:, None] == label[None, :]
    return float((B * same).sum() / (2.0 * m))


B_full, nodes_full, m_full = modularity_matrix(G_unweighted)
print(f"Full modularity matrix shape: {B_full.shape}")
print(f"Number of edges m = {m_full:.0f}")
print(f"B symmetric (Proposition 1):          {np.allclose(B_full, B_full.T)}")
print(f"||B 1|| (Proposition 2, should be ~0): {np.linalg.norm(B_full @ np.ones(len(nodes_full))):.2e}")

lam1, u1 = leading_eigenpair(B_full)
print(f"Leading eigenvalue lambda_1 = {lam1:.6f}")
print(f"Leading eigenvector (first 10 entries): {np.round(u1[:10], 4)}")
```

## 4. Spectral Bipartition of a Community

Given a set of nodes $C$, we:

1. Extract the restricted modularity matrix $B^{(C)}$
2. Compute its leading eigenpair $(\lambda_1^{(C)}, u_1^{(C)})$
3. If $\lambda_1^{(C)} \le 0$ declare $C$ indivisible
4. Otherwise split $C$ by the sign of $u_1^{(C)}$: $s_i=+1$ if $(u_1)_i>0$, else $-1$

**A practical subtlety (important).** For the plain block $B^{(C)}$ of a *cohesive* community, Rayleigh–Ritz gives $\lambda_1^{(C)} \ge \mathbf 1^\top B^{(C)}\mathbf 1/|C|$, and $\mathbf 1^\top B^{(C)}\mathbf 1>0$ whenever $C$ has more internal edges than chance. So for any community that is more cohesive than chance, $\lambda_1^{(C)}>0$ holds automatically, and the leading eigenvector then tends to be single-signed, i.e. the sign rule cannot separate any nodes (this is exactly what happens below). We therefore also stop when **all entries of $u_1^{(C)}$ have the same sign**, and we report the reason for stopping explicitly. Section 11 shows Newman's generalised matrix $B^{(g)}$, for which $\lambda_1\le 0$ is a genuinely valid stopping criterion.

```python
def spectral_bipartition(community, global_B, global_nodes, generalized=False, tol=1e-10):
    '''Try to bipartition `community` with the leading eigenvector of its restricted modularity matrix.

    Returns (status, C_plus, C_minus, lambda1) where status is one of
      'split'                   - a genuine bipartition was found
      'nonpositive_eigenvalue'  - lambda_1 <= 0  (handout stopping rule)
      'one_signed_eigenvector'  - lambda_1 > 0 but u_1 has a single sign, so the sign rule cannot split
      'too_small'               - fewer than 2 nodes
    '''
    community = sorted(community)
    if len(community) < 2:
        return 'too_small', [], [], 0.0

    node_to_idx = {n: i for i, n in enumerate(global_nodes)}
    idx = [node_to_idx[n] for n in community]
    B_C = global_B[np.ix_(idx, idx)]
    if generalized:                                   # Newman's B^(g): subtract diag(row sums)  (used in Section 11 only)
        B_C = B_C - np.diag(B_C.sum(axis=1))

    lam1, u1 = leading_eigenpair(B_C)

    if lam1 <= tol:
        return 'nonpositive_eigenvalue', [], [], float(lam1)

    C_plus  = [v for v, x in zip(community, u1) if x > 0]
    C_minus = [v for v, x in zip(community, u1) if x <= 0]
    if len(C_plus) == 0 or len(C_minus) == 0:
        return 'one_signed_eigenvector', [], [], float(lam1)

    return 'split', C_plus, C_minus, float(lam1)
```

## 5. Recursive Spectral Modularity Partitioning

We implement the classic recursive-bisection scheme:

1. Start with the full vertex set $V$.
2. Try to split a community (Section 4); if it splits, push both halves back onto the work list.
3. Record every intermediate partition (`history`) so that we can visualise the evolution and compute node metrics after each split.

To make the colours comparable between figures, one child of a split **keeps the parent's colour** and the other child gets a **new** colour.

```python
def recursive_spectral_modularity(G, generalized=False, verbose=True, max_splits=50):
    '''Recursive spectral bisection. Returns history, final_communities, B, nodes, m.'''
    B, nodes, m = modularity_matrix(G)

    active = [sorted(nodes)]
    final = []
    colour_of = {v: 0 for v in nodes}      # node -> colour id (stable across iterations)
    next_colour = 1
    iteration = 0

    history = [{
        'iteration': 0,
        'partition': [sorted(nodes)],
        'colours': dict(colour_of),
        'split': None,
        'lambda': None,
        'n_communities': 1
    }]

    while active and iteration < max_splits:
        C = active.pop(0)
        status, Cp, Cm, lam = spectral_bipartition(C, B, nodes, generalized=generalized)

        if status == 'split':
            iteration += 1
            for v in Cm:                                  # Cp keeps the parent's colour, Cm gets a new one
                colour_of[v] = next_colour
            next_colour += 1

            partition = final + active + [sorted(Cp), sorted(Cm)]
            history.append({
                'iteration': iteration,
                'partition': [list(c) for c in partition],
                'colours': dict(colour_of),
                'split': (list(C), list(Cp), list(Cm)),
                'lambda': lam,
                'n_communities': len(partition)
            })
            active.append(sorted(Cp))
            active.append(sorted(Cm))
            if verbose:
                print(f"Iteration {iteration}: split community of size {len(C)} "
                      f"(lambda_1 = {lam:.4f}) -> {len(Cp)} + {len(Cm)}")
        else:
            final.append(C)
            if verbose:
                reason = {'nonpositive_eigenvalue': f"lambda_1 = {lam:.2e} <= 0",
                          'one_signed_eigenvector': f"lambda_1 = {lam:.4f} > 0 but leading eigenvector has a single sign",
                          'too_small': "single node"}[status]
                print(f"    community of size {len(C):2d} is indivisible ({reason})")

    final.extend(active)
    return history, [sorted(c) for c in final], B, nodes, m


print("=" * 78)
print("Running recursive spectral modularity partitioning")
print("=" * 78)
history, final_comms, B_global, nodes_global, m = recursive_spectral_modularity(G_unweighted)
print("\n" + "=" * 78)
print(f"Finished after {history[-1]['iteration']} successful splits.")
print(f"Final number of communities: {len(final_comms)}")
for i, c in enumerate(sorted(final_comms, key=lambda x: -len(x)), 1):
    print(f"  Community {i} (size {len(c):2d}): {c}")
```

## 7. Node Metrics after Each Split

After every split (and for the initial whole graph, iteration 0) we compute four NetworkX metrics for **every node**:

- **Degree centrality** $C_D(i) = k_i/(n-1)$ — `nx.degree_centrality`
- **Betweenness centrality** $C_B(i) = \sum_{s\neq i\neq t}\sigma_{st}(i)/\sigma_{st}$ — `nx.betweenness_centrality`
- **Closeness centrality** $C_C(i) = (n-1)/\sum_{j\neq i}d(i,j)$ — `nx.closeness_centrality`
- **Clustering coefficient** $C(i) = 2T(i)/[k_i(k_i-1)]$ — `nx.clustering`

**Design choice.** If the metrics were always computed on the same full graph they would never change, and there would be no "evolution" to plot. So at each iteration every community $C$ is treated as its own network: we evaluate the metrics on the induced subgraph $G[C]$, and each node receives the value it has **inside its own community**. Hence a metric measures *how central the node is within its current community*, and it changes whenever a split removes inter-community edges and shrinks the community. At iteration 0 the single community is the whole graph, so iteration 0 equals the usual global metrics.

*Conventions:* a node alone in its community has no neighbours, so all four metrics are set to 0 (NetworkX would otherwise return 1.0 for the degree centrality of a one-node graph). Some communities produced by sign-splitting are not connected; for those, `nx.closeness_centrality` uses its default Wasserman–Faust correction, which scales by the fraction of nodes that are reachable.

```python
METRIC_FUNCS = {
    'degree':      nx.degree_centrality,
    'betweenness': nx.betweenness_centrality,
    'closeness':   nx.closeness_centrality,
    'clustering':  nx.clustering,
}

def metrics_for_partition(G, partition):
    '''Metrics of every node, each evaluated inside its own community G[C].'''
    out = {name: {} for name in METRIC_FUNCS}
    for comm in partition:
        if len(comm) == 1:
            for name in METRIC_FUNCS:
                out[name][comm[0]] = 0.0
            continue
        H = G.subgraph(comm)
        for name, f in METRIC_FUNCS.items():
            out[name].update(f(H))
    return out


# metric_history[metric][iteration][node] -> value
metric_history = {name: {} for name in METRIC_FUNCS}
for stage in history:
    vals = metrics_for_partition(G_unweighted, stage['partition'])
    for name in METRIC_FUNCS:
        metric_history[name][stage['iteration']] = vals[name]

iterations = sorted(metric_history['degree'].keys())
print("Metrics recorded for iterations:", iterations)

# sanity check: iteration 0 must coincide with NetworkX on the whole graph
for name, f in METRIC_FUNCS.items():
    whole = f(G_unweighted)
    assert all(abs(whole[v] - metric_history[name][0][v]) < 1e-12 for v in G_unweighted.nodes())
print("Check passed: iteration 0 == metrics on the whole graph.")

global_deg, global_bet = metric_history['degree'][0], metric_history['betweenness'][0]
print("\nTop-5 nodes by (global) degree centrality:")
for v, x in sorted(global_deg.items(), key=lambda t: -t[1])[:5]:
    print(f"  Node {v:2d}: {x:.4f}")
print("\nTop-5 nodes by (global) betweenness centrality:")
for v, x in sorted(global_bet.items(), key=lambda t: -t[1])[:5]:
    print(f"  Node {v:2d}: {x:.4f}")
```

## 9. Final Communities, Modularity Score & Comparison with the Real Split

We evaluate each partition with our own formula $Q=\frac{1}{2m}\sum_{ij}B_{ij}\delta(c_i,c_j)$ and cross-check it against `nx.community.modularity`. For a bipartition this equals $Q=\frac{1}{4m}s^\top B s$ (because $B\mathbf 1=0$).

```python
def nx_modularity(G, communities):
    return nx.community.modularity(G, [set(c) for c in communities], weight=None)

# --- modularity after every iteration ---------------------------------------
print("Modularity after each iteration")
Q_history = []
for stage in history:
    q_own = modularity_of_partition(B_global, m, nodes_global, stage['partition'])
    q_nx  = nx_modularity(G_unweighted, stage['partition'])
    assert abs(q_own - q_nx) < 1e-10
    Q_history.append(q_own)
    print(f"  iteration {stage['iteration']}: {stage['n_communities']} communities, Q = {q_own:.4f}   (NetworkX: {q_nx:.4f})")

Q_final = Q_history[-1]
gt_partition = [mr_hi, officer]
Q_gt = modularity_of_partition(B_global, m, nodes_global, gt_partition)
print(f"\nModularity of the final spectral partition ({len(final_comms)} communities): Q = {Q_final:.4f}")
print(f"Modularity of the historical ground-truth split (2 factions):        Q = {Q_gt:.4f}")

# --- Eq. (1) check for the first bipartition --------------------------------
first_parent, first_plus, first_minus = history[1]['split']
s = np.array([1.0 if v in set(first_plus) else -1.0 for v in nodes_global])
print(f"\nEq.(1) for the first split: s^T B s / (4m) = {s @ B_global @ s / (4 * m):.4f}")

# --- does the FIRST split recover the real fission? -------------------------
side_of_0 = first_plus if 0 in first_plus else first_minus
side_of_33 = first_minus if 0 in first_plus else first_plus
mis = sorted((set(side_of_0) - set(mr_hi)) | (set(side_of_33) - set(officer)))
print("\nFirst spectral bipartition vs. what actually happened:")
print(f"  side containing node 0  ({len(side_of_0)} nodes): {sorted(side_of_0)}")
print(f"  side containing node 33 ({len(side_of_33)} nodes): {sorted(side_of_33)}")
print(f"  misclassified nodes: {mis}  ->  {len(G_unweighted) - len(mis)}/{len(G_unweighted)} correct "
      f"({100 * (len(G_unweighted) - len(mis)) / len(G_unweighted):.1f}%)")
```

### How does community structure influence each metric?
- **Degree centrality** only counts edges inside the community and is divided by $|C|-1$. It rises for almost every node because the denominator shrinks faster than friends are lost, and the two leaders reach 1.0. The exceptions are nodes that lose many friends to the other side: node 23 (0.294 → 0.250 when its neighbours 25 and 27 are split off), node 9 (0.061 → 0.059) and node 11 (0.030 → 0.067 → 0.000, because its only friend, node 0, ends up in another community).
- **Closeness** behaves like degree: distances shrink once the long inter-community paths are removed, so values rise towards 1 for hubs of small, dense communities. It is the metric most affected by isolation (node 11: 0.367 → 0.500 → 0).
- **Betweenness** is the most community-sensitive metric. It concentrates on nodes that tie sub-groups of the *same* community together (node 0: 0.438 → 0.690 inside the 16-node Mr. Hi side → 0.354 once the periphery $\{4,5,6,10,11,16\}$ is split off), and it vanishes for nodes whose brokerage was *between* communities (nodes 2 and 8). In other words, global betweenness largely measures who links communities; within-community betweenness measures who holds a community together.
- **Clustering coefficient** is the only metric where hubs are *not* on top: globally the highest values (1.0) belong to low-degree nodes such as 7, 12, 14, 15, 16, whose few friends all know each other, while the two leaders have low values (node 0: 0.150, node 33: 0.110) since they connect people who do not know each other. Removing inter-community edges raises the clustering of nodes that had links across the divide (node 2: 0.244 → 0.900, node 8: 0.500 → 1.000), because their remaining neighbours all belong to one tight group.

### Caveats & limitations
- After iteration 0, metrics are evaluated *inside* communities, so values from communities of different sizes are not directly comparable (a node in a 5-node community can reach 1.0 more easily than one in a 13-node community). The relative ordering inside a community is the meaningful quantity.
- Sign-splitting does not guarantee connected communities: the 6-node community $\{4,5,6,10,11,16\}$ present at iterations 2–3 is not connected (node 11's only friend is node 0, which sits in another community), and it is then split into $\{4,5,6,10,16\}$ and the singleton $\{11\}$. The generalised $B^{(g)}$ of Section 11 keeps node 11 together with node 0 and avoids the singleton.
- Node 8 cannot be "fixed" by modularity-based refinement: moving it to Mr. Hi's side *lowers* $Q$ from 0.3715 to 0.3582 (the real split differs from the spectral one only in node 8). It is a case where the friendship structure and the actual decision disagree.
- The method is deterministic and parameter-free; for large graphs the dense `eigh` should be replaced by a sparse Lanczos eigensolver.


## 11. Optional Extension – Why the Handout's Stopping Rule Needs a Correction

In Newman (2006) the matrix used to split a group $g$ is the *generalised* modularity matrix

$$
B^{(g)}_{ij} \;=\; B_{ij} \;-\; \delta_{ij}\sum_{l\in g} B_{il},
$$

i.e. the block $B^{(C)}$ with its row sums subtracted from the diagonal. It satisfies $B^{(g)}\mathbf 1=\mathbf 0$ (like the global $B$), so $\lambda_1\ge 0$ always, and the modularity gain of splitting $g$ with $s\in\{\pm1\}^{|g|}$ is **exactly** $\Delta Q = \frac{1}{4m}s^\top B^{(g)}s$. With the plain block $B^{(C)}$ this identity does **not** hold. The cell below checks this numerically for the two communities created by the first split.

```python
def delta_Q_check(parent, other):
    '''Split `parent` with the leading eigenvector of (a) plain block, (b) Newman's B^(g); compare gains.'''
    idx = parent
    B_C = B_global[np.ix_(idx, idx)]
    B_g = B_C - np.diag(B_C.sum(axis=1))
    Q_before = modularity_of_partition(B_global, m, nodes_global, [parent, other])
    rows = []
    for label, M in [('plain block B^(C)  (handout)', B_C), ("Newman's B^(g)", B_g)]:
        lam, u = leading_eigenpair(M)
        s = np.where(u > 0, 1.0, -1.0)
        c1 = [v for v, x in zip(parent, s) if x > 0]
        c2 = [v for v, x in zip(parent, s) if x < 0]
        dQ_true = modularity_of_partition(B_global, m, nodes_global, [c1, c2, other]) - Q_before
        rows.append((label, len(c1), len(c2), dQ_true, s @ B_C @ s / (4 * m), s @ B_g @ s / (4 * m)))
    return rows

C0, C33 = sorted(side_of_0), sorted(side_of_33)
for name, parent, other in [("community containing node 0", C0, C33), ("community containing node 33", C33, C0)]:
    print(f"Splitting the {name} (size {len(parent)}):")
    print(f"  {'matrix used to find s':32s} {'sizes':>7s} {'true dQ':>9s} {'s^T B^(C) s/4m':>15s} {'s^T B^(g) s/4m':>15s}")
    for label, a, b, dq, f_plain, f_gen in delta_Q_check(parent, other):
        print(f"  {label:32s} {a:>3d}+{b:<3d} {dq:9.4f} {f_plain:15.4f} {f_gen:15.4f}")
    print()
```
